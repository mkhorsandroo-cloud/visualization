In this session, we explored how to choose the most appropriate data visualization for a given context, emphasizing that data visualizations are never fully neutral. 

Through comparisons of different visualizations of similar datasets (such as gun violence and unemployment data), we learned how design choices can shape interpretation and emotional response without necessarily being dishonest. 

The session introduced the concepts of aesthetic, substantive, and perceptual qualities of visualizations, highlighting how multiple visualizations can be equally accurate yet perceived very differently. 

We also discussed how intended purpose, audience, and medium influence visualization choices, and how cognitive principles—such as Gestalt laws and cognitive load—affect how viewers understand charts. 

Overall, the session reinforced that effective data visualization is a rhetorical and interpretive act, requiring deliberate, ethical design decisions to balance clarity, honesty, and communication goals.
