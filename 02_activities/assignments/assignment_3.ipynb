Data Visualization Using Toronto Major Crime Indicators

Dataset Description and Source

The dataset used in this project is Major Crime Indicators (MCI) from the City of Toronto Open Data Portal, published by the Toronto Police Service. The dataset contains event-level records of major crimes in Toronto, including assault, robbery, break and enter, auto theft, and theft from vehicle.

Although the data was accessed through a cleaned version hosted on Kaggle for ease of use, the original data source is the City of Toronto Open Data Portal, which satisfies the dataset requirements of the assignment.

Each row in the dataset represents a single crime occurrence, with attributes such as:

    REPORT_YEAR

    MCI_CATEGORY

    DIVISION

    NEIGHBOURHOOD

    Temporal and spatial metadata

No personally identifiable information is included.

Visualization 1: Trends of Major Crimes Over Time

Software Used:

This visualization was created using Python in a Jupyter Notebook environment. The primary libraries used were:

    pandas for data manipulation

    matplotlib and seaborn for visualization


The intended audience includes:

    Toronto residents

    Urban planners and policy makers

    Journalists and researchers

    Students studying urban issues or public safety

Message and Purpose:

The purpose of this visualization is to examine how major crime occurrences in Toronto have changed over time. By grouping crime events by reporting year, the visualization highlights long-term trends and temporal fluctuations in crime levels.

This visualization communicates that crime patterns evolve over time and that understanding these changes is essential for evidence-based decision-making, policy development, and public discourse.

Design Considerations:

    A line chart was chosen because it effectively communicates trends over time.

    The x-axis represents reporting year, and the y-axis represents the number of crime occurrences.

    Clear axis labels and a descriptive title were used to minimize ambiguity.

    The scale was chosen carefully to avoid exaggerating year-to-year differences.

    Visual clutter was minimized to focus attention on the overall trend.

Reproducibility:

This visualization is fully reproducible. All steps—from loading the dataset to grouping, counting, and plotting—are implemented in Python code. The dataset is publicly available, and the analysis can be replicated by running the provided notebook.

Accessibility:

The visualization does not rely solely on color to convey information.

Font sizes and line thickness were selected to enhance readability.

Labels and titles provide sufficient context for interpretation by assistive technologies.

Visualization 2: Distribution of Major Crime Types in Toronto

Software Used:

The same Python-based workflow was used for this visualization to maintain consistency and reproducibility.


This visualization is intended for:

    Community organizations

    City officials

    Policy makers

    Members of the general public interested in crime patterns

Message and Purpose:

This visualization compares the relative frequency of different major crime categories in Toronto. By aggregating individual crime events by category, the chart highlights which crime types occur most frequently and which are less common.

The goal is to encourage data-driven discussion about resource allocation, prevention strategies, and public awareness, rather than focusing on isolated incidents.

Design Considerations:

    A horizontal bar chart was selected to improve readability of category labels.

    Crime categories are sorted by frequency to support intuitive comparison.

    Neutral colors were used to avoid sensationalizing the data.

    Spacing and layout were optimized to reduce cognitive load.

Reproducibility:

As with the first visualization, this chart is fully reproducible using the provided Python code and the publicly available dataset.

Accessibility:

The chart uses a colorblind-safe palette.

Labels are clear and descriptive.

Ordering of bars supports quick visual comprehension.

Ethical Considerations and Impacted Communities:

Crime data has the potential to reinforce stereotypes or stigmatize communities if not handled carefully. This project uses aggregated data and avoids naming or ranking specific neighbourhoods in a way that could encourage misinterpretation.

The individuals and communities most impacted by crime—and by policy decisions informed by crime data—include marginalized and underserved populations. These visualizations were designed to emphasize systemic patterns rather than individual blame, supporting responsible and ethical data communication.

Feature Selection and Data Scope:

Only high-level variables relevant to temporal and categorical analysis were included, such as reporting year and crime category. Variables that could increase the risk of identification or misinterpretation were excluded.

This approach balances analytical usefulness with ethical responsibility.

Underwater Labour:

Significant “underwater labour” contributed to the final product, including:

Understanding the structure of an event-level dataset

Cleaning column names and validating data types

Choosing appropriate aggregation methods (counting events rather than summing values)

Iteratively testing visualization designs

Refining layout, labels, and explanatory text

These invisible steps were essential to producing accurate, ethical, and accessible visualizations.

Conclusion:

This project demonstrates how publicly available municipal data can be transformed into meaningful and responsible visual narratives. By combining reproducible methods, thoughtful design choices, and ethical considerations, the visualizations support informed discussion about crime trends in Toronto and illustrate the value of data visualization in public-interest contexts.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv(
    "major-crime-indicators.csv"
)

# Visualization 1: Crime trends over time
crime_trend = df.groupby("REPORT_YEAR").size()

plt.figure(figsize=(8,4))
crime_trend.plot()
plt.title("Major Crime Trends in Toronto Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Occurrences")
plt.tight_layout()
plt.show()

# Visualization 2: Distribution of crime types
crime_types = df.groupby("MCI_CATEGORY").size().sort_values()

plt.figure(figsize=(8,4))
crime_types.plot(kind="barh")
plt.title("Distribution of Major Crime Types in Toronto")
plt.xlabel("Number of Occurrences")
plt.ylabel("Crime Type")
plt.tight_layout()
plt.show()
```
